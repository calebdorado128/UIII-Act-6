## Realizando simulacion de inteligencia artificial con if, if else, elif y while
# Problema numero 15: Pago de servicios

```python
# --- SIMULACIÓN DE ASISTENTE DE PAGO DE SERVICIOS (IA) ---

# Configuración inicial de datos (Simulando una base de datos)
nombre_cliente = "Alex"
saldo_cuenta = 3000.0
pin_secreto = 4321

# Deudas pendientes de los servicios
deuda_luz = 450.0
deuda_agua = 120.0
deuda_internet = 350.0

print("IA: ¡Hola! Soy tu asistente virtual de pagos. Detecto un intento de acceso.")
print("IA: Por seguridad, necesito que te identifiques.")

# Entrada interactiva para el PIN
pin_ingresado = int(input("CLIENTE: Angresa tu PIN de 4 dígitos: "))

# Verificación de identidad con if-else
if pin_ingresado == pin_secreto:
    print(f"\nIA: ¡Acceso concedido! Bienvenido/a de nuevo, {nombre_cliente}.")

    # Usamos un número (1) en lugar de True para controlar el ciclo
    estado_asistente = 1

    # El ciclo while continuará mientras estado_asistente sea igual a 1
    while estado_asistente == 1:
        print("\n==========================================")
        print(f"IA: Estado de tu cuenta: ${saldo_cuenta} disponibles.")
        print("IA: ¿Qué servicio deseas gestionar hoy?")
        print("Opciones: 1. Ver deudas | 2. Pagar Luz | 3. Pagar Agua | 4. Pagar Internet | 5. Salir")
        print("==========================================")

        opcion = input("CLIENTE (Elige una opción del 1 al 5): ")

        # Estructura elif para la toma de decisiones múltiple
        if opcion == "1":
            print("\nIA: Consultando tus recibos pendientes...")
            print(f"- Recibo de Luz: ${deuda_luz}")
            print(f"- Recibo de Agua: ${deuda_agua}")
            print(f"- Recibo de Internet: ${deuda_internet}")

        elif opcion == "2":
            if deuda_luz == 0:
                print("\nIA: Tu servicio de Luz ya se encuentra pagado. ¡Buen trabajo!")
            else:
                print(f"\nIA: Tu deuda de Luz es de ${deuda_luz}.")
                confirmar = input("IA: ¿Deseas pagarla ahora mismo? (si/no): ")

                if confirmar == "si" or confirmar == "SI":
                    if saldo_cuenta >= deuda_luz and deuda_luz > 0:
                        saldo_cuenta = saldo_cuenta - deuda_luz
                        print(f"IA: ¡Pago exitoso! Se han descontado ${deuda_luz} de tu cuenta.")
                        deuda_luz = 0.0  # La deuda se limpia
                    else:
                        print("IA: Lo siento. No tienes saldo suficiente para cubrir este pago.")
                else:
                    print("IA: Operación cancelada. El recibo seguirá pendiente.")

        elif opcion == "3":
            if deuda_agua == 0:
                print("\nIA: Tu servicio de Agua ya está al corriente.")
            else:
                print(f"\nIA: Tu deuda de Agua es de ${deuda_agua}.")
                confirmar = input("IA: ¿Deseas realizar el pago? (si/no): ")

                if confirmar == "si" or confirmar == "SI":
                    if saldo_cuenta >= deuda_agua and deuda_agua > 0:
                        saldo_cuenta = saldo_cuenta - deuda_agua
                        print(f"IA: ¡Pago procesado! Se han descontado ${deuda_agua} de tu saldo.")
                        deuda_agua = 0.0
                    else:
                        print("IA: Fondos insuficientes para pagar el Agua.")
                else:
                    print("IA: Entendido. No se realizó ningún cargo.")

        elif opcion == "4":
            if deuda_internet == 0:
                print("\nIA: Tu Internet está totalmente pagado.")
            else:
                print(f"\nIA: Tu deuda de Internet es de ${deuda_internet}.")
                confirmar = input("IA: ¿Procedemos con el pago? (si/no): ")

                if confirmar == "si" or confirmar == "SI":
                    if saldo_cuenta >= deuda_internet and deuda_internet > 0:
                        saldo_cuenta = saldo_cuenta - deuda_internet
                        print(f"IA: ¡Éxito! Tu servicio de Internet está pagado.")
                        deuda_internet = 0.0
                    else:
                        print("IA: No cuentas con dinero suficiente en tu saldo actual.")
                else:
                    print("IA: Pago de internet pospuesto.")

        elif opcion == "5":
            print("\nIA: Cerrando sesión de forma segura...")
            # Cambiamos el valor a 0 para que la condición del 'while' deje de cumplirse
            estado_asistente = 0

        else:
            print("\nIA: No logré entender esa opción. Por favor, introduce un número del 1 al 5.")

else:
    print("\nIA: Alerta. El PIN es incorrecto. Acceso bloqueado por seguridad.")

print("\nIA: Gracias por utilizar el Asistente Inteligente de Pagos. ¡Hasta pronto!")
```
